## 6. Clean text columns
First find the columns that have extra spaces at the start or end.

```python
all_tables = {
    "employee": employee,
    "performance": performance,
    "education_level": education_level,
    "rating_level": rating_level,
    "satisfied_level": satisfied_level,
}

for name, df in all_tables.items():
    for col in df.select_dtypes(include=["object", "string"]).columns:
        n_bad = (df[col] != df[col].str.strip()).sum()
        if n_bad > 0:
            print(f"{name}.{col}: {n_bad} values with extra spaces")
```

Now remove the extra spaces.

```python
for name, df in all_tables.items():
    for col in df.select_dtypes(include=["object", "string"]).columns:
        df[col] = df[col].str.strip()

print("Done: extra spaces removed")
```

Check again (nothing should be printed).

```python
for name, df in all_tables.items():
    for col in df.select_dtypes(include=["object", "string"]).columns:
        n_bad = (df[col] != df[col].str.strip()).sum()
        if n_bad > 0:
            print(f"{name}.{col}: still {n_bad} values with extra spaces")
print("Check finished")
```

## 7. Fix date columns
`HireDate` is written as year-month-day. `ReviewDate` is written as month/day/year.

```python
employee["HireDate"] = pd.to_datetime(employee["HireDate"], format="%Y-%m-%d")
performance["ReviewDate"] = pd.to_datetime(performance["ReviewDate"], format="%m/%d/%Y")

print(employee["HireDate"].dtype, performance["ReviewDate"].dtype)
print("Hire dates:", employee["HireDate"].min().date(), "to", employee["HireDate"].max().date())
print("Review dates:", performance["ReviewDate"].min().date(), "to", performance["ReviewDate"].max().date())
```

## 8. Check the links between tables
These checks protect us from wrong results later when we join the tables.

```python
# Do all reviews belong to a known employee?
orphans = set(performance["EmployeeID"]) - set(employee["EmployeeID"])
print("Reviews with unknown employee:", len(orphans))

# How many employees have at least one review?
with_reviews = performance["EmployeeID"].nunique()
print("Employees with reviews:", with_reviews, "of", len(employee))
print("Employees without any review:", len(employee) - with_reviews)
```

```python
# Do the numeric codes match the lookup tables?
print("Education codes in employee:", sorted(employee["Education"].unique().tolist()))
print("Codes in education_level:", sorted(education_level["EducationLevelID"].unique().tolist()))
print()
for col in ["EnvironmentSatisfaction", "JobSatisfaction", "RelationshipSatisfaction", "WorkLifeBalance"]:
    print(col, sorted(performance[col].unique().tolist()))
for col in ["SelfRating", "ManagerRating"]:
    print(col, sorted(performance[col].unique().tolist()))
```

## 9. Save the clean files

```python
employee.to_csv("../data/clean/employee_clean.csv", index=False)
performance.to_csv("../data/clean/performance_clean.csv", index=False)
education_level.to_csv("../data/clean/education_level_clean.csv", index=False)
rating_level.to_csv("../data/clean/rating_level_clean.csv", index=False)
satisfied_level.to_csv("../data/clean/satisfied_level_clean.csv", index=False)

print("All clean files saved to data/clean")
```
